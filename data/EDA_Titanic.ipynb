## Insights & Conclusion

### Key Insights

1. **Category-wise Sales:**  
   Electronics category generated the highest total sales, while Stationery had the lowest sales.

2. **Product-wise Profit:**  
   Laptop generated the highest total profit, followed by Monitor and Office Chair.

3. **Region-wise Profit:**  
   The North region generated the highest total profit, followed by West, South, and East.

4. **Sales vs Profit:**  
   Total Sales were ₹4,412,300 while Total Profit was ₹801,690.40.

5. **Top 5 Products:**  
   Laptop, Monitor, Office Chair, Desk, and Headphones were the top 5 products based on total profit.

### Conclusion

The analysis shows that Electronics and Laptop products are the strongest contributors to sales and profit. The North region is the most profitable region. These insights can help the business focus on high-performing products and regions and improve the performance of lower-performing categories and products.

```python
df["Date"] = pd.to_datetime(df["Date"])

daily_sales = df.groupby("Date")["Sales"].sum()

daily_sales
```

```python
daily_sales.plot(
    kind="line",
    figsize=(12, 5)
)

plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales (₹)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
